# Performing descriptive statistic and visulaizatoin on final_cleaned_data

```python
import os
import pandas as pd

# --- Path Config ---
BASE_DIR = os.path.abspath(os.path.join(os.getcwd(), ".."))
DATA_DIR = os.path.join(BASE_DIR, "data")
OUTPUTS_DIR = os.path.join(BASE_DIR, "outputs")
STATS_DIR = os.path.join(OUTPUTS_DIR, "statistics")
PLOTS_DIR = os.path.join(OUTPUTS_DIR, "plots")
METRICS_DIR = os.path.join(OUTPUTS_DIR, "metrics")

# Make sure subfolders exist
os.makedirs(STATS_DIR, exist_ok=True)
os.makedirs(PLOTS_DIR, exist_ok=True)
os.makedirs(METRICS_DIR, exist_ok=True)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the cleaned, transformed dataset
df = pd.read_csv(os.path.join(STATS_DIR, "final_cleaned_data.csv"))

# Initialize list to store descriptive stats
desc_stats = []

# Iterate over each column
for col in df.columns:
    # Compute frequency and proportion
    freq = df[col].value_counts(dropna=False)
    prop = df[col].value_counts(normalize=True, dropna=False) * 100
    mode = df[col].mode().iloc[0] if not df[col].mode().empty else None

    for level in freq.index:
        desc_stats.append({
            "Feature": col,
            "Level": level,
            "Frequency": freq[level],
            "Proportion (%)": round(prop[level], 2),
            "First Mode": mode if level == mode else ""
        })

# Convert to DataFrame
summary_df = pd.DataFrame(desc_stats)

# Save summary to CSV
summary_df.to_csv(os.path.join(STATS_DIR, "categorical_summary_part6.csv"), index=False)
print("Descriptive statistics saved to 'categorical_summary_part6.csv'")
```

# Visualization

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

# Create bar charts for each categorical variable
for col in df.columns:
    plt.figure(figsize=(8, 5))

    order = df[col].value_counts().index  # Order bars by frequency
    
    # Use 'hue=col' to avoid the seaborn FutureWarning
    sns.countplot(data=df, y=col, order=order, hue=col, palette="pastel", legend=False)

    plt.title(f"Distribution of {col}")
    plt.xlabel("Count")
    plt.ylabel("Category")
    plt.tight_layout()

    # Clean file name: remove special characters and spaces
    safe_filename = col.replace("/", "-").replace("+", "plus").replace(" ", "_")

    # Create subdirectory for plots
    part6_plots_dir = os.path.join(PLOTS_DIR, "part6_preprocessing")
    os.makedirs(part6_plots_dir, exist_ok=True)

    # Save plot
    plt.savefig(os.path.join(part6_plots_dir, f"{safe_filename}_barplot.png"))

    plt.close()

print("Bar plots saved in 'plots/' folder.")
```
